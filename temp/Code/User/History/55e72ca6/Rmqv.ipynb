```python
import numpy as np
import pandas as pd

# Set random seed for reproducibility
np.random.seed(42)

# Generate synthetic data
n_customers = 500

customer_ids = np.arange(1000, 1000 + n_customers)
ages = np.random.randint(18, 70, size=n_customers)
genders = np.random.choice(['Male', 'Female'], size=n_customers)
annual_income = np.random.normal(50000, 15000, n_customers).astype(int)
spending_score = np.random.randint(1, 101, size=n_customers)
num_purchases = np.random.randint(1, 50, size=n_customers)
recency = np.random.randint(1, 365, size=n_customers)
online_vs_instore = np.random.rand(n_customers)
response_to_promotions = np.random.randint(1, 11, size=n_customers)

# Create a DataFrame
df = pd.DataFrame({
    'CustomerID': customer_ids,
    'Age': ages,
    'Gender': genders,
    'Annual Income': annual_income,
    'Spending Score': spending_score,
    'Number of Purchases': num_purchases,
    'Recency': recency,
    'Online vs. In-Store Purchases': online_vs_instore,
    'Response to Promotions': response_to_promotions
})

# Save the dataset to a CSV file
df.to_csv('customer_data.csv', index=False)

print("Dataset created and saved as 'customer_segmentation_data.csv'.")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score

# Load the dataset
df = pd.read_csv('customer_segmentation_data.csv')

# Encode the 'Gender' column
df['Gender'] = df['Gender'].map({'Male': 0, 'Female': 1})

# Select features for clustering
features = ['Age', 'Gender', 'Annual Income', 'Spending Score',
            'Number of Purchases', 'Recency',
            'Online vs. In-Store Purchases', 'Response to Promotions']

# Standardize the features
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df[features])

# Determine the optimal number of clusters
wcss = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, random_state=42)
    kmeans.fit(df_scaled)
    wcss.append(kmeans.inertia_)

# Plot the Elbow Method
plt.figure(figsize=(10, 5))
plt.plot(range(1, 11), wcss, marker='o')
plt.title('Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')
plt.show()

# Silhouette Score
for n_clusters in range(2, 11):
    kmeans = KMeans(n_clusters=n_clusters, random_state=42)
    cluster_labels = kmeans.fit_predict(df_scaled)
    silhouette_avg = silhouette_score(df_scaled, cluster_labels)
    print(f'For n_clusters = {n_clusters}, the silhouette score is {silhouette_avg:.3f}')

# Apply K-Means with optimal clusters
optimal_clusters = 4
kmeans = KMeans(n_clusters=optimal_clusters, random_state=42)
df['Cluster'] = kmeans.fit_predict(df_scaled)

# Apply PCA for visualization
pca = PCA(n_components=2)
df_pca = pca.fit_transform(df_scaled)

df['PCA1'] = df_pca[:, 0]
df['PCA2'] = df_pca[:, 1]

# Plot the clusters
plt.figure(figsize=(10, 7))
sns.scatterplot(x='PCA1', y='PCA2', hue='Cluster', data=df, palette='viridis')
plt.title('Customer Segments')
plt.show()

# Cluster interpretation
cluster_summary = df.groupby('Cluster')[features].mean()
print(cluster_summary)

# Suggested marketing strategies
for i in range(optimal_clusters):
    print(f"Cluster {i} Summary:")
    print(cluster_summary.iloc[i])
    print("Suggested Marketing Strategy:")
    print("-" * 40)
```
