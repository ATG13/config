```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Create sample DataFrame for pivot table
data = {'index': ['A', 'B', 'C', 'A', 'B', 'C'],
        'columns': ['X', 'Y', 'Z', 'X', 'Y', 'Z'],
        'values': [1, 2, 3, 4, 5, 6]}

df = pd.DataFrame(data)

# Create a pivot table
pivot_table = df.pivot(index='index', columns='columns', values='values')
print("Pivot Table:\n", pivot_table)

# # Pivot table with aggregation
# pivot_table = df.pivot_table(index='index', columns='columns', values='values', aggfunc='sum')
# print("Pivot Table with Aggregation:\n", pivot_table)

# # Pivot table with missing value handling
# pivot_table = df.pivot_table(index='index', columns='columns', values='values', fill_value=0)
# print("Pivot Table with Missing Value Handling:\n", pivot_table)

# # Pivot table with multiple aggregations
# pivot_table = df.pivot_table(index='index', columns='columns', values='values', aggfunc=['sum', 'mean'])
# print("Pivot Table with Multiple Aggregations:\n", pivot_table)
```
