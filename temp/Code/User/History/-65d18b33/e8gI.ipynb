```python
import pandas as pd
dataset=pd.read_csv("auto-mpg.csv")
dataset.head()
```

```python
# Min-Max Normalization
dataset1=dataset.copy(deep=True)
dataset1["mpg"]=(dataset1["mpg"]-dataset1["mpg"].min())/(dataset1["mpg"].max()-dataset1["mpg"].min())
dataset1["displacement"]=(dataset1["displacement"]-dataset1["displacement"].min())/(dataset1["displacement"].max()-dataset1["displacement"].min())
dataset1["horsepower"]=(dataset1["horsepower"]-dataset1["horsepower"].min())/(dataset1["horsepower"].max()-dataset1["horsepower"].min())
dataset1["acceleration"]=(dataset1["acceleration"]-dataset1["acceleration"].min())/(dataset1["acceleration"].max()-dataset1["acceleration"].min())
dataset1["weight"]=(dataset1["weight"]-dataset1["weight"].min())/(dataset1["weight"].max()-dataset1["weight"].min())
```

```python
a=dataset1["mpg"]
b=dataset1["acceleration"]
c=dataset1["weight"]
d=dataset1["horsepower"]
e=dataset1["displacement"]
vals={"mpg":a,"acceleration":b,"weight":c,"horsepower":d,"displacement":e}
mmnormalizedvalues=pd.DataFrame(vals)
mmnormalizedvalues
```

```python
dataset2=dataset.copy(deep=True)
dataset2["mpg"]=(dataset2["mpg"]-dataset2["mpg"].mean())/dataset2["mpg"].std()
dataset2["acceleration"]=(dataset2["acceleration"]-dataset2["acceleration"].mean())/dataset2["acceleration"].std()
dataset2["weight"]=(dataset2["weight"]-dataset2["weight"].mean())/dataset2["weight"].std()
dataset2["horsepower"]=(dataset2["horsepower"]-dataset2["horsepower"].mean())/dataset2["horsepower"].std()
dataset2["displacement"]=(dataset2["displacement"]-dataset2["displacement"].mean())/dataset2["displacement"].std()

a1=dataset2["mpg"]
b1=dataset2["acceleration"]
c1=dataset2["weight"]
d1=dataset2["horsepower"]
e1=dataset2["displacement"]
vals2={"mpg":a1,"acceleration":b1,"weight":c1,"horsepower":d1,"displacement":e1}
zscorevalues=pd.DataFrame(vals2)
zscorevalues
```
